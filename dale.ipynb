```python
import requests 
import json
```

```python
def chamada(message):

    messages = [
        {
            "role": "user",
            "content": message
        }
    ]

    url = "http://localhost:8081/v1/chat/completions"

    headers={
        "Content-Type": "application/json"
    }

    payload = {
        "model" : "qwen",
        "messages": messages,
        "stream": False
    }

    response = requests.post(url=url, headers=headers, json=payload)

    data = response.json()

    return data["choices"][0]["message"]["content"]
```

```python
def chamada1(message, model_id):

    messages = [
        {
            "role": "user",
            "content": message
        }
    ]

    url = "http://10.30.4.153:8082/v1/chat/completions"

    headers = {
        "Content-Type": "application/json"
    }

    payload = {
        "model": model_id,
        "messages": messages,
        "stream": False
    }

    response = requests.post(url=url, headers=headers, json=payload)

    print("Status: ", response.status_code)
    print("Body: ", response.text)

    data = response.json()

    return data["choices"][0]["message"]["content"]
```
