```python
import json
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def parse_json_data(path_to_json, output_file):
    
    # Opens the JSON file
    with open(path_to_json,'r') as infile:
        lst_dicts = json.load(infile)

    # Parses the JSON data to a dataframe
    d = {}
    for elem in lst_dicts:
        # Extracts the ID from inside each JSON entry
        id = elem['id']
        d[id] = elem
        # Parses the MODEL-QUANTIZATION string
        model_str = d[id]['model']
        model_str_quant = model_str.split('-')[-1]
        model_str_model = model_str.replace(model_str_quant, '')[:-1]
        d[id]['quantization'] = model_str_quant
        d[id]['llm'] = model_str_model

        # Expands the internal dictionary 'tokens'
        for k in d[id]['tokens']:
            d[id][k] = d[id]['tokens'][k]

    # List of columns that we are interested in
    lst_keep = ['quantization', 'llm', 'duration_ms', 'tokens_per_second']

    # Converts the temporary dictionary d to a dataframe
    df = pd.DataFrame.from_dict(d, orient='index')
    df = df.reset_index(drop=True)
    df = df[lst_keep]
    df = df.rename({'llm':'model'}, axis=1)
    df['duration_min'] = df['duration_ms']/(1000*60)
    df = df.drop(['duration_ms'], axis=1)

    df.to_csv(output_file, index=False)

    return df
```

```python
parse_json_data(path_to_json='../task_1/all_metrics_clean.json', output_file='raw_data_task1.csv')
```
